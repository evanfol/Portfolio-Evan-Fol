```python
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, FunctionTransformer
```

# I- Import des données

```python
df = pd.read_csv("billets_production.csv", sep=",")
df.head()
```

# II- Préparation des données

```python
# Garder l'identifiant à part
ids = df["id"]
ids
```

```python
df.info()
```

# III- Création de variables

```python
#L'aire
df["Area"] = round(df["length"] * ((df["height_right"] + df["height_left"])/2),2)
```

```python
#Différence des tailles
df["diff_height"] = abs(round(df["height_right"] - df["height_left"],2))
```

```python
# Différence des marges
df["diff_margin"] = abs(round(df["margin_up"] - df["margin_low"],2))
```

```python
#Somme des marges
df["sum_margin"] = round(df["margin_up"] + df["margin_low"],2)
```

```python
# Charger modèle
modele_charge = joblib.load("modele_final.joblib")

# Variables
X = df.drop(columns=["id"]) # Enlever id des variables utilisées par le modèle


# Prédictions
predictions = modele_charge.predict(X)
probas = modele_charge.predict_proba(X)[:, 1]

resultats = pd.DataFrame({
    "id": ids,
    "prediction": predictions,
    "probabilite_faux": probas,
    "resultat": ["FAUX ❌" if p == 1 else "VRAI ✅" for p in predictions]
})

print(resultats)
```
